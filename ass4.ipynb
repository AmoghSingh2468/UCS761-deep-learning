```python
import numpy as np
import pandas as pd

# Load dataset
data = pd.read_csv("abalone.data", header=None)

# Drop first column (Sex: categorical)
data = data.iloc[:,1:]

# Features and target
X = data.iloc[:,:-1].values
y = data.iloc[:,-1].values.reshape(-1,1)

# Normalize features
X = (X - X.mean(axis=0)) / X.std(axis=0)

# Network parameters
input_neurons = X.shape[1]
hidden_neurons = 10
output_neurons = 1

learning_rate = 0.01
epochs = 1000

# Initialize weights
W1 = np.random.randn(input_neurons, hidden_neurons)
b1 = np.zeros((1, hidden_neurons))

W2 = np.random.randn(hidden_neurons, output_neurons)
b2 = np.zeros((1, output_neurons))

# Sigmoid function
def sigmoid(x):
    return 1/(1 + np.exp(-x))

def sigmoid_derivative(x):
    return x*(1-x)

# Training
for epoch in range(epochs):

    # Forward propagation
    hidden_input = np.dot(X, W1) + b1
    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output, W2) + b2
    y_pred = sigmoid(final_input)

    # Error
    error = y - y_pred

    # Backpropagation
    d_output = error * sigmoid_derivative(y_pred)

    error_hidden = np.dot(d_output, W2.T)
    d_hidden = error_hidden * sigmoid_derivative(hidden_output)

    # Update weights
    W2 += learning_rate * np.dot(hidden_output.T, d_output)
    b2 += learning_rate * np.sum(d_output, axis=0, keepdims=True)

    W1 += learning_rate * np.dot(X.T, d_hidden)
    b1 += learning_rate * np.sum(d_hidden, axis=0, keepdims=True)

# Prediction
print("Sample Predictions:\n")

for i in range(5):
    hidden = sigmoid(np.dot(X[i], W1) + b1)
    output = sigmoid(np.dot(hidden, W2) + b2)
    print("Actual Rings:", y[i][0], "Predicted:", output[0][0])
```
