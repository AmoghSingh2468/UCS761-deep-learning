```python
import numpy as np

def perceptron(x1, x2, w1, w2, b):
    net = w1*x1 + w2*x2 + b
    if net >= 0:
        return 1
    else:
        return 0

inputs = [(0,0),(0,1),(1,0),(1,1)]

print("AND Gate")
for x1,x2 in inputs:
    print(x1,x2, perceptron(x1,x2,1,1,-1.5))

print("\nOR Gate")
for x1,x2 in inputs:
    print(x1,x2, perceptron(x1,x2,1,1,-0.5))

print("\nNAND Gate")
for x1,x2 in inputs:
    print(x1,x2, perceptron(x1,x2,-2,-2,3))

print("\nNOR Gate")
for x1,x2 in inputs:
    print(x1,x2, perceptron(x1,x2,-1,-1,0.5))
```
